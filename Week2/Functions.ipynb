```python
def cal_marks_summary(marks):
  total = sum(marks)
  avg = total/len(marks)
  highest = max(marks)
  lowest = min(marks)
  return total, avg, highest, lowest
```

```python
student_marks = [88,98,10.5, 56, 75, 99]
```

```python
type(student_marks)
```

```python
total, avg, highest, lowest = cal_marks_summary(student_marks)
```

```python
total, highest, lowest = cal_marks_summary(student_marks)
```

```python
def show_profiles(**details):#variable length keyword arguments
  for key, value in details.items():
    print(f"{key}:{value}")
```

```python
show_profiles(name="neha", age=24, city="bengaluru")
```

```python
show_profiles(name="neha", age=24, city="bengaluru", enrolled=True)
```
