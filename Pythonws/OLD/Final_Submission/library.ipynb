```python
# read csv book data for library management system
def read_csv(file_path):
    import csv
    books = []
    with open(file_path, mode='r', newline='', encoding='utf-8') as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            books.append(row)
    return books

# print the book data in a formatted way
def print_books(books):
    for book in books:
        print(f"Title: {book['title']}, Author: {book['author']}, Year: {book['genre']}, ISBN: {book['ISBN']}")  


# main function to read and print book data
def main():
    file_path = 'book.csv'  # specify the path to your CSV file
    books = read_csv(file_path)
    books
    print_books(books)

2+1
```
