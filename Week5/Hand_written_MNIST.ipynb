```python
#(n, h, w, c) #RGB
import numpy as np
xtrain = xtrain.reshape(xtrain.shape[0], np.prod(xtrain.shape[1:]))
xtest = xtest.reshape(len(xtest), np.prod(xtest.shape[1:]))
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
784*0.6
```

```python
450*0.6
```

```python
250*0.6
```

```python
model = Sequential()
model.add(Dense(450, input_shape=(784,), activation="relu"))
model.add(Dense(250, activation="relu"))
model.add(Dense(150,  activation="relu"))
model.add(Dense(50,  activation="relu"))
model.add(Dense(10, activation="softmax"))  #0-9
```

```python
from tensorflow.keras.utils import to_categorical
ytrain = to_categorical(ytrain, num_classes=10)
ytest = to_categorical(ytest)
```

```python
ytrain[0]
```

```python
#256*128+128  y=mx+c
```

```python
model.summary()
```

```python
model.compile(loss="categorical_crossentropy",
              optimizer="adam",
              metrics=["accuracy"])
```

```python
ytrain[12]
```

```python
60000/64
```

```python
%%time
h1 = model.fit(xtrain, ytrain, epochs=10, batch_size=64, validation_data=(xtest, ytest))
```

```python
ypred = model.predict(xtest)
```

```python
np.round(ypred[0],2)
```

```python
np.argmax(np.round(ypred[0],2))
```

```python
ypred[0]
```

```python
pred = np.argmax(ypred, axis=1)
```

```python
actual = np.argmax(ytest, axis=1)
```

```python
from sklearn.metrics import accuracy_score, classification_report
print(classification_report(actual, pred))
accuracy_score(actual, pred)
```

```python

plt.plot(h1.history["accuracy"])
plt.plot(h1.history["val_accuracy"])
```
